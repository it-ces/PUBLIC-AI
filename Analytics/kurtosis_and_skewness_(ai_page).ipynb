```python
import pandas as pd
import numpy as np

x_normal  = np.random.normal(loc = 4, scale = 3, size = 1000)
x_uniform = np.random.uniform(1, 100, 1000)
x_exponential = np.random.exponential(8, 1000)

df = pd.DataFrame({'normal':x_normal,
                   'uniform':x_uniform,
                   'exponential':x_exponential})
```

```python
df.describe()
```

```python
df['uniform'].hist()
```

Positive skewness (+): los datos se concentran principalmente en valores bajos y existe una cola larga hacia los valores altos.



Negative skewness (−): los datos se concentran principalmente en valores altos y existe una cola larga hacia los valores bajos.



Zero skewness (0): no existe asimetría medida por el tercer momento, aunque eso no garantiza que la distribución sea simétrica


La curtosis (kurtosis) indica principalmente qué tan pesadas son las colas de una distribución, es decir, qué tanta tendencia tiene a producir valores extremos. refleja la importancia relativa de los valores extremos respecto a su variabilidad. si $X \sim N$ entonces $K_{normal}=3$,

\begin{equation}
K = K_{var} - 3
\end{equation}

```python
df.skew()
```

```python
df.kurtosis()
```
